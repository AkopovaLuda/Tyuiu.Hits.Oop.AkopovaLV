<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Person в C#, который будет представлять информацию о
человеке. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу Person:
• Атрибуты: Имя (Name), Возраст (Age), Пол (Gender).
• Методы:
o GetInfo(): метод для получения информации о человеке в виде строки.
o SayHello(): метод для вывода приветствия от человека.
Требования к производным классам:
1. Студент (Student): Должен содержать дополнительные атрибуты, такие как
Университет (University) и Курс (Course). Метод SayHello() должен быть
переопределен для добавления информации о курсе при приветствии.
2. Работник (Employee): Должен содержать дополнительные атрибуты, такие
как Компания (Company) и Зарплата (Salary). Метод GetInfo() должен быть
переопределен для включения информации о компании и зарплате.
3. Преподаватель (Teacher) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Предмет (Subject) и Стаж работы
(Experience). Метод SayHello() должен быть переопределен для добавления
информации о преподаваемом предмете при приветствии.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```

// интерфейсы для множественного наследования 
public interface IWorker
{
    string Company { get; set; }
    double Salary { get; set; }
    void Work();
}

public interface IResearcher
{
    string ResearchTopic { get; set; }
    void ConductResearch();
}


//  базовый класс 3+ атрибута и метода
public class Person
{
    public string Name { get; set; }
    public int Age { get; set; }
    public string Gender { get; set; }
    public string Email { get; set; } // Новый атрибут

    public Person() { }

    public Person(string name, int age, string gender, string email)
    {
        Name = name;
        Age = age;
        Gender = gender;
        Email = email;
    }

    public virtual void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}.");
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age}, Пол: {Gender}, Email: {Email}");
    }

    public void UpdateEmail(string newEmail)
    {
        Email = newEmail;
        Console.WriteLine($"{Name} обновил email на {Email}");
    }
}


// простое наследование Student от Person 
public class Student : Person
{
    public string University { get; set; }
    public int Course { get; set; }
    public double AverageGrade { get; set; } // Новый атрибут
    public bool IsInCampus { get; set; }      // Новый атрибут

    public Student() : base() { }

    public Student(string name, int age, string gender, string email, string university, int course, double averageGrade, bool isInCampus) 
        : base(name, age, gender, email)
    {
        University = university;
        Course = course;
        AverageGrade = averageGrade;
        IsInCampus = isInCampus;
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Я студент {Name}, учимся в {University} на {Course} курсе.");
    }

    public void Study()
    {
        Console.WriteLine($"{Name} сидит на лекции и грызет гранит науки.");
    }

    public void PassExam()
    {
        Console.WriteLine($"{Name} успешно сдал сессию со средним баллом {AverageGrade}!");
    }
}


// Сложное наследование
// Цепочка: Person -> Student -> GraduateStudent 
public class GraduateStudent : Student
{
    public string ThesisTopic { get; set; }  // Тема диссертации
    public int PublicationsCount { get; set; } // Количество научных статей

    public GraduateStudent(string name, int age, string gender, string email, string university, int course, double averageGrade, bool isInCampus, string thesisTopic, int publicationsCount) 
        : base(name, age, gender, email, university, course, averageGrade, isInCampus)
    {
        ThesisTopic = thesisTopic;
        PublicationsCount = publicationsCount;
    }

    public override void GetInfo()
    {
        base.GetInfo(); // Выведет инфо от родителя Student и дедушки Person
        Console.WriteLine($"Аспирант. Тема работы: '{ThesisTopic}', Статей: {PublicationsCount}");
    }

    public void WriteThesis()
    {
        Console.WriteLine($"{Name} пишет кандидатскую диссертацию на тему: '{ThesisTopic}'.");
    }
}


// Множественное наследование
// Класс WorkingStudent наследуется от Student И реализует интерфейс IWorker
public class WorkingStudent : Student, IWorker
{
    public string Company { get; set; }
    public double Salary { get; set; }

    public WorkingStudent(string name, int age, string gender, string email, string university, int course, double averageGrade, bool isInCampus, string company, double salary) 
        : base(name, age, gender, email, university, course, averageGrade, isInCampus)
    {
        Company = company;
        Salary = salary;
    }

    public void Work()
    {
        Console.WriteLine($"{Name} совмещает учебу в {University} и работу в компании {Company} с зарплатой {Salary} руб.");
    }
}




Student student = new Student("Алексей", 20, "Мужской", "alex@mail.ru", "МГУ", 2, 4.7, true);
student.SayHello();
student.PassExam();

GraduateStudent gradStudent = new GraduateStudent("Елена", 24, "Женский", "elena@mail.ru", "СПбГУ", 5, 5.0, false, "Искусственный интеллект", 4);
gradStudent.SayHello();
gradStudent.GetInfo(); // Отработает цепочка base() до самого верха
gradStudent.WriteThesis();

WorkingStudent workStudent = new WorkingStudent("Максим", 21, "Мужской", "max@yandex.ru", "ТюмГУ", 3, 4.2, true, "Яндекс", 95000);
workStudent.SayHello();
workStudent.Work();

```
